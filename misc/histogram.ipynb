```python
import os
import csv
import pandas as pd
from matplotlib import pyplot as plt
from collections import defaultdict, Counter
from itertools import combinations
import numpy as np
```

```python
df = pd.read_csv("D:/Bao/Document/_VGU/_Semester 5/Projects/[TI+NR]/[TI+NR]evaluation_results_20251119_224437.csv")
```

```python
df.head(5)
```

```python
# Histogram
counts, bins, patches = plt.hist(df["answer_relevancy"], bins=[i/10 for i in range(1,10)], edgecolor='black')

# Add labels at bottom (bin count) and top (total frequency)
for i in range(len(counts)):
    # Coordinates for text placement
    x = (bins[i] + bins[i+1]) / 2  # middle of each bar
    y_top = counts[i]
    
    # Show frequency at top (0.3 is offset to slightly move the text away from the bar edge)
    plt.text(x, y_top + 0.3, str(int(y_top)), ha='center', va='bottom', fontsize=8, color='black')

plt.xlabel("Count values")
plt.ylabel("Frequency")
plt.title("Answer Relevancy of of MedGemma 4b")
plt.tight_layout()
plt.show()
```

```python
# Histogram
counts, bins, patches = plt.hist(df["answer_correctness"], bins=[i/100 for i in range(-1,10)], edgecolor='black')

# Add labels at bottom (bin count) and top (total frequency)
for i in range(len(counts)):
    # Coordinates for text placement
    x = (bins[i] + bins[i+1]) / 2  # middle of each bar
    y_top = counts[i]
    
    # Show frequency at top (0.3 is offset to slightly move the text away from the bar edge)
    plt.text(x, y_top + 0.3, str(int(y_top)), ha='center', va='bottom', fontsize=8, color='black')

plt.xlabel("Count values")
plt.ylabel("Frequency")
plt.title("Answer Correctness of MedGemma 4b")
plt.tight_layout()
plt.show()
```
